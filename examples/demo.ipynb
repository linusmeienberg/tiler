## Library usage (no widget)

You can also build a fundamental domain and render its covering programmatically, e.g. to export an SVG.

```python
import numpy as np
from tiler import FundamentalDomain, generate_covering, svg_from_paths
from IPython.display import SVG, display

domain = FundamentalDomain(group="p4g", cell_size=1.0)
domain.add_path(np.array([[0.05, 0.05], [0.3, 0.1], [0.15, 0.35]]))

covering = generate_covering(domain, viewport_min=(-2, -2), viewport_max=(2, 2))
svg = svg_from_paths(covering, viewport_min=(-2, -2), viewport_max=(2, 2))
display(SVG(svg))
```
